```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing   import StandardScaler
from sklearn.linear_model    import LogisticRegression
from sklearn.tree            import DecisionTreeClassifier 
```

```python
wine_data = pd.read_csv('datasets/winequality-white.csv')

print(wine_data.head())
print(wine_data.shape)
```

```python
x_wine = wine_data.drop(columns= ['quality', 'alcohol'])  ## 종속변수가 될 quality 칼럼과 str이 섞인 alcohol 칼럼 제거
print(x_wine.head())                            ## inplace = False가 기본값이고, quality 칼럼이 제거된 DataFrame을 반환함
```

```python
y_wine = wine_data['quality']
print(y_wine.head())
```

```python
scaler = StandardScaler()   ## 정규화 스케일러
```

```python
x_wine_stdsc = scaler.fit_transform(x_wine) ## 와인 데이터의 독립변수들 정규화

print(x_wine_stdsc)
```

```python
x_tr, x_tst, y_tr, y_tst = train_test_split(x_wine_stdsc, y_wine, test_size = 0.3, random_state= 46)


# from sklearn.model_selection import train_test_split
# train_test_split() 함수는 독립변수와 종속변수를 나누어서 인자로 집어넣으면
# 차례대로 독립변수의 train set, test set 그리고 종속변수의 train set, test set을 반환하는 함수이다

# test_size는 test set의 비율이고
# random_state는 이 코드를 다시 실행했을 때 처음 나눈 dataset 상태로 돌아오도록 save 해주는 기능이다.

# 두 매개변수는 반드시 나온다고 생각하자!
```

```python
print('x_학습set size:', x_tr.shape)
print('x_testset size:', x_tst.shape)
print('y_학습set size:', y_tr.shape)
print('y_testset size:', y_tst.shape)
```
