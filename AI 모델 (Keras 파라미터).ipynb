### Keras 딥러닝 파라미터 알아보기

#### 1. Sequential

```python
'''
Sequential은 모델의 뼈대로, layer를 순차적으로 쌓아 올리는 
가장 단순한 형태의 딥러닝 모델 구조

기본 코드 구조
    model = Sequential()
    model.add(Dense(units, activation=None, input_shape=None))
    model.add(Dropout(rate))
    model.add(BatchNormalization())

주요 등장 파라미터
    Dence
        units
        레이어의 출력 뉴런 개수 (출력벡터 차원)
        
        Activation (활성화 함수)
        출력에 적용할 비선형 함수 (relu, sigmoid, softmax)
        
        input_shape
        첫 층에서의 입력 데이터 차원 (튜플로 지정)
        ex) input_shape = (X.shape[1], 0)

    Dropout
        학습할 때 랜덤으로 뉴런 일부를 꺼버림(출력 0)
        특정 뉴런에 의존하지 않게 해 과적합 방지
    
    BatchNormalize
    미니배치 단위로 입력 값을 정규화 (평균 0, 분산 1 근처)
    정규화 후 다시 학습 가능한 scale, shift 파라미터 적용 
        기울기 소실 및 폭주 문제를 완화함  
'''
```

#### 2. compile

```python
'''
compile은 Sequential에서 정의된 모델의 
학습 방법 (optimization process)을 정의하는 단계

기본 코드 구조
    model.compile(optimizer=None, loss=None, metrics=[None])
    
주요 등장 파라미터

    optimizer(최적화 알고리즘)
        가중치를 어떻게 업데이트할지 결정하는 방법 (adam, sgd)

    loss(손실 함수)
        모델이 얼마나 잘못 예측했는지 측정하는 함수
        회귀는 mse, 이진분류는 binary_crossentropy, 다중분류는 categorial_crossentropy

    metrics(평가 지표)
        학습 과정을 모니터링 하기 위한 지표 
        손실함수와 별개 - accuracy, mae, mse 등
        metrics는 사용시 꼭 [] 안에 작성해야 함 - list 형태로 받기 때문
        
    loss_weight
        다중 출력 모델일 때 출력치를 조정할 가중치
'''
```

#### 3. fit()

```python
'''
모델의 학습을 당담하는 fit

훈련 데이터(X, y)를 넣어서 
지정한 epochs와 batch 단위 등으로 모델 파라미터를 업데이트

기본 코드 구조
    history = model.fit(
        X_train, y_train,
        batch_size= None,
        epochs= None,
        verbose= None,
        validation_split= 0.2,
    또는validation_data(X_valid, y_valid),
        shffle=True,
        callbacks = [
        EarlyStopping(monitor='val_loss', patience=5),
        ModelCheckpoint('best_model.h5', save_best_only=True)
        ]
    )

파라미터

    X,y
        학습할 입력 데이터와 레이블

    batch_size
        한번에 학습할 데이터의 수 (기본값 32)

    epochs
        전체 데이터셋을 몇 번 반복 학습할지
        
    verbose
        학습 로그 출력 형태 (0 없음, 1 진행바, 2 에포크별 출력)

    validation_split 또는 validation_data
        검증 데이터 설정시 훈련 데이터 일부 자동 분리

    shuffle
        epoch마다 데이터 순서를 섞을지 여부
'''
```

#### 4. fit() 가운데 callbacks

```python
'''
callbacks
    딥러닝 학습fit() 도중에 특정 이벤트 (에포크 끝/배치 끝 등)이 발생했을 때
    자동으로 실행되는 함수나 객체


    
EarlyStopping
    딥러닝 모델 학습 중 성능(val_loss, val_accuracy)이 
    더 이상 개선되지 않으면 학습을 조기 종료하는 콜백
    아래 es를 callbacks = [] 속에 넣으면 된다.
    
    es = EarlyStopping(
        monitor= 'val_loss',         # 감시할 지표 (val_loss 사용시 validation_data 사용해야함)
        patience=5,                  # 개선이 없더라도 기다릴 epoch 수
        mode= 'min',                 # min(값이 작을수록 좋음)/ max(값이 커질수록 좋음)
        restore_best_weights= True   # True 면 최고 성능을 낸 가중치로 복원
        )



ModelCheckpoint
    최고 성능의 가중치 저장
    학습 도중, 조건에 맞는 모델(가중치)를 자동으로 저장해주는 콜백
    
    mc = ModelCheckpoint(
        filepath= 'best_model.h5',      # 저장할 경로와 파일 이름 (확장자는 .h5 / .keras)
        monitor= 'val_loss',
        save_best_only= True,           # True : 성능이 개선될 때만 / False : 매 epoch
        save_weights_only= False        # True : 가중치만 저장  / False : 모델 구조와 가중치 모두
        mode= 'min'
    )

ReduceLROnPlateau - 성능 정체 시 learning rate 줄이기

TensorBoard/CSVLogger - 훈련 과정 기록
'''
```
