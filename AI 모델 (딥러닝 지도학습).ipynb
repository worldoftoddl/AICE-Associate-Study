### 1. 인공신경망

```python
'''
이번 시간에 배울 내용
    1. 인공신경망 구조
    2. 가중치와 편향
    3. 활성화 함수
'''
```

```python
'''
1-1. 인공신경망 구조

인간의 뇌 구조(뉴런)를 본따 만든 기계학습 알고리즘
지도학습이므로 입력과 출력을 번갈아 보며 규칙을 스스로 찾는 구조.

    뉴런과 노드

        뉴런은 인공신경망에서 데이터를 처리하는 연단 단위(기능적 구분)
            입력 - 가중치 계산 - 합산 - 활성화 - 출력의 기능을 수행함

        노드는 신경망 내에서 뉴런이 위치한 지점(구조적 구분)
            각 뉴런이 노드에 위치해 계산한 출력값을 다음 뉴런으로 전달함
            
    다시 말해, 뉴런(신경세포)은 특정 연산을 수행하는 주체, 
            노드(세포 구조 속 위치)는 그 연산이 일어나는 위치!


    인겅신경망의 구조

            인공신경망은 크게 입력층, 은닉층, 출력층으로 구성됨
                입력층 : 데이터를 입력받는 layer (뉴런 수는 데이터 차원 수에 따라 결정)
                은닉층 : 입력 데이터의 특성을 추출
                출력층 : 결과를 내보내는 layer (뉴런 수는 결과 형태 - 회귀 or 분류에 따라 결정)
'''
```

```python
'''
1-2. 가중치와 편향

    가중치(Weight)

        입력 값이 결과에 얼마나 영향을 미치는지 결정하는 숫자
        특정 입력에 대한 가중치가 클수록 결과에 더 큰 영향
        학습 과정은 이 가중치를 반복적으로 조정해가며 더 나은 예측을 찾는 과정
        

    편향 (Bias)

        뉴런의 계산 결과에 항상 더해지는 기본값
            입력이 0이어도 출력은 0이 되지 않음 : 뉴런이 완전히 꺼지지 않도록 (가장 중요!!)
            출력 범위를 넓히기 위해             : 다양한 상황에 적용 가능
            더 유연한 학습을 위해               : 모델의 표현력을 높힐 수 있음
'''
```

```python
'''
1-3. 활성화함수

뉴런이 계산한 값을 그대로 출력하지 않고,
복잡한 문제를 풀기 위해 바꾸고,
신경망이 학습할 수 있는 기울기도 만들어주는 함수

    
    역할

        비선형성 추가(중요): 학습곡선이나 효용함수의 기울기가 왜 기우는가? 세상은 선형이 아니다.
        출력 값 범위 조정  : 출력값이 과도하게 커지는 것을 제한하여 안정적 학습을 유도
        계층적 특징 학습   : 은닉층을 통해 의미있는 중간 표현을 학습
        신경망 표현력 증가 : 단일 계층으로 해결할 수 없는 문제를 다층구조로 학습함
        
        
    종류

        1. 시그모이드 함수 (입력을 0과 1로 바꿔줌)
                σ(x) = 1/(1 + e^(-x))
                출력 범위: (0, 1)
            출력 값이 확률처럼 해석될 수 있어 이진 분류 문제에서 활용됨
            입력값이 너무 작거나 크면 기울기 소실(출력이 0과 1에 수렴) 문제가 발생할 수 있음
            
        2. 하이퍼볼릭 탄젠트 (입력을 -1부터 1까지로 바꿔줌)
                tanh(x) = (e^(2x) - 1)/(e^(2x) + 1)
                출력 범위: (-1, 1)
            0을 중심으로 대칭이기 때문에 신경망이 양수/음수 정보를 더 잘 표현할 수 있음
            마찬가지로 입력값이 극단적인 경우 기울기 소실(출력이 1과 -1에 수렴)이 발생함
            지수함수를 사용해 연산 비용이 상대적으로 높음.
        
        3. 렐루 (Rectified Linear Unit, ReLU)
                ReLU(x) = max(0, x) = {
                x,  if x > 0
                0,  if x ≤ 0
                }
                출력 범위: [0, +∞)
            게산이 빨라 가장 많이 사용됨
            입력이 양수일 때는 그대로, 음수일때는 0으로 만드는 함수
            헷징처럼 하방이 0으로 확정(희소성 제공)
            죽은 ReLU 문제가 발생해 뉴런이 작동하지 않을 수 있음
            입력이 매우 클 경우 기울기 폭주 문제로 학습이 불안정할 가능성
            
        4. 리키 렐루 (Leaky ReLU)
                LeakyReLU(x) = max(αx, x) = {
                x,   if x > 0
                αx,  if x ≤ 0
                }
                α는 작은 상수 (보통 0.01)
                출력 범위: (-∞, +∞)
            렐루 함수의 죽은 ReLU 문제를 해결하기 위해
            음수를 아주 작게 출력시켜 뉴런이 죽지 않음
            일반적으로 0.01 정도로 작은 정도로 출력시키며 아주 작은 출력과 기울기를 유지함
            
        5. 소프트맥스 (Softmax)
            모델이 예측한 각 클래스의 점수를 확률로 반환하는 함수
            다중 클래스 분류 문제에서 마지막 출력층의 활성화 함수로 사용
            분류 모델에서 사용된다.

    정리
        선택 가이드
            깊은 네트워크: ReLU 계열 필수
            이진 분류: 출력층에 Sigmoid
            다중 분류: 출력층에 Softmax
            음수 정보 중요: Leaky ReLU 고려
            전통적 얕은 네트워크: Tanh도 괜찮음

        함수        주 용도           요약
        Sigmoid     이진 분류         Y/N
        Tanh        -/0/+ 모두        감정 분석
        ReLU        대부분의 은닉층   이미지/음성
        Leaky ReLU  개선된 ReLU       미세 반응(이상) 감지     
        Softmax     다층 분류(출력층) 숫자/감정 분류
'''
```
