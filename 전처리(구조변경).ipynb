# 데이터 전처리 - 구조변경 : 그룹화, 집계, 데이터병합과 추가

```python
import pandas as pd

# 데이터 로드
file_path = "datasets/heart.csv"
df = pd.read_csv(file_path)

# 컬럼명 변경
df.columns = ['나이', '성별', '흉통유형', '안정시혈압', '콜레스테롤', '공복혈당', '심전도',
              '최대심박수', '운동유발협심증', 'ST우울증', '운동후ST경사', '주요혈관수',
              '결함유형', '심장질환여부']

# 데이터 확인
print("변경된 데이터프레임 컬럼명:")
print(df.columns)
print("\n데이터 상위 5개 행:")
print(df.head())
```

## 1. 그룹화 및 집계

```python
'''
그룹화(Grouping)는 데이터를 특정 기준에 따라 묶는 작업이고
집계(Aggregation)은 묶인 데이터를 요약하는 작업이다.
이를 통해 평균, 합계, 개수등의 통계를 도출할 수 있다.

    groupby()
        pandas 라이브러리의 메서드로, 데이터를 특정 열 기준으로 그룹화한다
    
    grouped.agg()
        마찬가지로 pandas 라이브러리의 메서드로, 여러 집계 함수를 동시에 적용한다
'''


'''
그룹화(groupby) → 인덱스(그래프의 왼쪽) 모양 결정
집계(agg) → 컬럼(그래프의 위쪽) 모양 결정

그룹화는 "어떻게 나눌까?"(인덱스), 집계는 "무엇을 계산할까?"(컬럼)
'''
```

```python
# 1. 성별 기준 데이터 그룹화
grouped = df.groupby('성별')

print('성별 그룹화된 데이터:')
print(grouped.size())        ## groupby()는 데이터프레임이 아닌 시리즈를 반환함

print('그룹별 평균')
print(grouped.mean())
```

```python
# 2. 그룹화된 데이터 집계
grouped_stats = grouped.agg({'나이' : ['max', 'min'],
                             '콜레스테롤' : ['mean', 'std']
                            })

print(grouped_stats)        ## 보이듯이 멀티 칼럼 형식의 데이터를 만들어준다
```

```python
# 3. 성별과 흉통유형을 기준으로 그룹화
grouped_multi = df.groupby(['성별', '흉통유형']).mean()     ## 두 가지 기준 그루핑 : 멀티 인덱스

print("\n성별과 흉통유형별 주요 열의 평균:")
print(grouped_multi)            ## 마찬가지로 멀티 인덱스 형식의 데이터를 보여준다
```

## 2. 피벗테이블

```python
'''
피봇테이블은 데이터를 특정 기준에 따라 요약하거나 변환하는데 사용하는 도구이다.
pandas 라이브러리에서 제공하는 기능은 엑셀과 유사하며,
데이터를 행, 열, 값으로 구성하여 다양한 방식으로 요약할 수 있다.

    df.pivot_table()     
'''
```

```python
import pandas as pd

# 데이터 로드
file_path = "datasets/heart.csv"
df = pd.read_csv(file_path)

# 컬럼명 변경
df.columns = ['나이', '성별', '흉통유형', '안정시혈압', '콜레스테롤', '공복혈당', '심전도',
              '최대심박수', '운동유발협심증', 'ST우울증', '운동후ST경사', '주요혈관수',
              '결함유형', '심장질환여부']
```

```python
# 피벗테이블 생성
pivot = df.pivot_table(index= '성별', columns='흉통유형', values= '최대심박수', aggfunc= ['mean', 'max'])
'''
index: 그래프 왼쪽 (그루핑)
column: 그래프 위쪽 (집계)
values: 계산 대상 값들
aggfunc: values에 행할 계산

따라서 위의 코드는 종을 성별, 횡을 흉통유형으로 둔 채로 최대 심박수의 평균과 최댓값을 계산하는 코드이다
'''
## 위의 모든 항목들 [] 리스트 형식으로 묶어 멀티인덱스나 멀티 칼럼도 가능함

print('####성별과 흉통유형별 평균 최대 심박수####\n')
print(pivot)
```

```python
pivot_multi_index = df.pivot_table(index = ['성별', '흉통유형'], values= '최대심박수', aggfunc= ['mean', 'max'])

print(pivot_multi_index)
```

## 4. 데이터 병합과 추가

### 4-1. 데이터 병합

```python
'''
동일한 칼럼을 가지고 있는 두 개의 데이터를 병합한다.

    merge()
        on= 공통 칼럼이며, 병합의 기준이 되는 칼럼이다.
        how= 병합 방식으로, 'inner' 'outer' 'left' 'right' 'cross'등이 가능하다. 
            'inner'
            두 데이터의 공통 칼럼 중 교집합만 가져옴
            'outer'
            두 데이터의 합집합. 모두 가져옴
            'left'
            왼쪽 데이터(첫번째)에 오른쪽 데이터를 붙임. 왼쪽 데이터는 모든 행을 유지함
            'right'
            오른쪽 데이터(두번째)에 오른쪽 데이터를 붙임. 오른쪽 데이터는 모든 행을 유지함
'''
```

```python
import pandas as pd

# 첫 번째 데이터프레임 생성
data1 = {
    '제품ID': [1, 2, 3],
    '제품명': ['노트북', '모니터', '키보드']
}
df1 = pd.DataFrame(data1)

# 두 번째 데이터프레임 생성
data2 = {
    '제품ID': [1, 2, 4],
    '가격': [1000, 200, 50]
}
df2 = pd.DataFrame(data2)

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
```

#### 4-1-1. 내부 조인과 외부 조인

```python
# 내부 조인 - 교집합
inner_join = pd.merge(df1, df2, on='제품ID', how='inner')

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
print("\n내부 조인 결과:")
print(inner_join)
```

```python
# 외부 조인 - 합집합
outer_join = pd.merge(df1, df2, on='제품ID', how='outer')

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
print("\n외부 조인 결과:")
print(outer_join)
```

#### 4-1-2. 왼쪽 조인과 오른쪽 조인

```python
# 왼쪽 조인 - 첫번째에 두번째 붙이기
left_join = pd.merge(df1, df2, on='제품ID', how='left')

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
print("\n왼쪽 조인 결과:")
print(left_join)
```

```python
# 오른쪽 조인 - 두번째에 첫번째 붙이기
right_join = pd.merge(df1, df2, on='제품ID', how='right')

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
print("\n오른쪽 조인 결과:")
print(right_join)
```

#### 4-1-3. 교차 조인

```python
# 교차 조인 
cross_join = pd.merge(df1, df2, how='cross') 

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
print("\n교차 조인 결과:") 
print(cross_join)
```

### 4-2. 데이터 추가

```python
'''
데이터 병합(join)이 다른 열을 가진 데이터를 합치는 과정이라면,
데이터 추가(union)은 동일한 열을 가진 데이터를 세로로 확장하는 작업이다.

데이터 추가에서는 concat()을 사용할 것이다.
'''
```

```python
# 첫 번째 데이터프레임 생성
data1 = {
    '이름': ['홍길동', '이순신'],
    '나이': [30, 40]
}
df1 = pd.DataFrame(data1)

# 두 번째 데이터프레임 생성
data2 = {
    '이름': ['강감찬', '유관순'],
    '나이': [50, 20]
}
df2 = pd.DataFrame(data2)

# 데이터프레임 출력
print("첫 번째 데이터프레임:")
print(df1)
print("\n두 번째 데이터프레임:")
print(df2)
```

#### 4-2-1. 같은 column을 가진 데이터의 결합(union)

```python
union_df = pd.concat([df1, df2], ignore_index=True)         ## ignore_index : 기존 데이터프레임 왼쪽의 인덱스 (0,1,2~)를 무시할건지

print("## 첫번째 df")
print(df1)

print('\n## 두번째 df')
print(df2)

print('\n## 결합된 df')
print(union_df)
```

#### 4-2-2. 다른 column을 가진 데이터의 결합

```python
# 열이 다른 데이터프레임 생성
data3 = {
    '이름': ['정약용', '김유신'],
    '주소': ['서울', '경주']
}
df3 = pd.DataFrame(data3)

# 데이터프레임 결합
union_with_different_columns = pd.concat([df1, df2, df3], ignore_index=True)

print("## 첫번째 df")
print(df1)

print('\n## 두번째 df')
print(df2)

print('\n## 세번째 df')
print(df3)

print("\n열이 다른 데이터프레임 결합 결과:")
print(union_with_different_columns)
```

## 5. 데이터 파생 - 변수(column) 생성

```python
import pandas as pd
import numpy as np

# 샘플 데이터 생성
df = pd.DataFrame({
    '이름' : ['철수','영희','민수','지수'],
    '나이' : [23, 35, 45, 52],
    '월급' : [250, 400, 600, 700]
})

# 연령대 변수 생성
df['연령대'] = pd.cut(df['나이'], bins= [0,30,40,50,100],
                   labels= ['20대', '30대', '40대', '50대 이상'])

# 누적 월급 계산
df['누적 월급'] = df['월급'].cumsum()

# 편차 계산
df['월급_편차'] = df['월급'] - df['월급'].mean()

# 로그 변환 (로그화)
df['월급_로그화'] = np.log1p(df['월급'])

print(df)
```

```python
df = pd.DataFrame({
    '이름' : ['철수','영희','민수','지수'],
    '성별' : ['남','여','남','여'],
    '주소' : ['서울 강남구','부산 해운대구','서울 종로구','대전 서구']
})

# 성별을 코드로 변환
df['성별_코드'] = df['성별'].map({'여' : 0, '남' : 1})

# 주소에서 도시와 구 추출
df['도시'] = df['주소'].str.split(" ").str[0]
df['구'] = df['주소'].str.split(" ").str[1]

print(df)
```

```python
# 혹은 주소를 한 번에 분할해서 도시와 구 컬럼 생성
df[['도시', '구']] = df['주소'].str.split(" ", expand=True)

print(df)
```
