# 탐색적 데이터 분석 - 기초 통계

```python
'''
이번 회차는 탐색적 데이터 분석(Exploratory Data Analysis), EDA 이다.

데이터를 시각적으로 탐구하고 통계적으로 요약하여
데이터의 구조 특성, 패턴 및 이상치 등을 파악하는 과정이다.

EDA는 머신러닝 모델링 및 데이터 분석의 첫단계이며,
데이터를 잘 이해하고 분석의 방향을 설정하는 중요한 초석이다.
'''
```

```python
'''
기술통계 (Descriptive Statistics)
"데이터를 요약하고 설명하는 통계"

목적
수집한 데이터의 특성을 숫자나 그래프로 요약
데이터가 어떻게 생겼는지 한눈에 파악

주요 방법
중심경향: 평균, 중앙값, 최빈값
분산: 표준편차, 분산, 범위
분포형태: 히스토그램, 상자그림
관계: 상관계수, 산점도




추론통계 (Inferential Statistics)
"표본으로 모집단을 추측하는 통계"

목적
표본 데이터를 이용해 전체 모집단에 대한 결론 도출
가설 검증, 예측, 일반화

주요 방법
가설검정: t-검정, 카이제곱 검정
신뢰구간: 모집단 평균의 범위 추정
회귀분석: 변수들 간의 관계 예측
분산분석: 그룹 간 차이 검정


우리가 EBR에 적용할 방법은 기술통계이다.
'''
```

```python
# Pandas를 사용한 평균 계산
import pandas as pd

data = {'Score': [80, 90, 100, 70, 85]}  # 예제 데이터
df = pd.DataFrame(data)
print(df)
```

```python
# 평균 계산
mean_value = df['Score'].mean()
print("평균(Mean):", mean_value)

# 중앙값 계산
median_value = df['Score'].median()
print("\n중앙값(Median):", median_value)

# 최빈값 계산
mode_value = df['Score'].mode()
print("\n최빈값(Mode):", mode_value)

# 최댓값과 최솟값 계산
max_value = df['Score'].max()
min_value = df['Score'].min()
print("\n최댓값(Max):", max_value)
print("최솟값(Min):", min_value)

# 표준편차 계산
std_value = df['Score'].std()
print("\n표준편차(std):",std_value)

# 분산 계산
variance_value = df['Score'].var()
print("분산(Variance):", variance_value)
```

```python
# heart.csv 파일 읽기
file_path = 'datasets/heart.csv'  # CSV 파일 경로 지정
df = pd.read_csv(file_path)  # 데이터프레임 생성
```

```python
# 기초통계량 분석
print("\n기초통계량 분석 결과:\n")
print(df.describe())  # 전체 통계 요약
```

```python
# 평균(Mean) 계산
print("평균(Mean):")
print(df.mean())  # 각 열의 평균값 계산

# 중앙값(Median) 계산
print("\n중앙값(Median):")
print(df.median())  # 각 열의 중앙값 계산
```

```python
# 최빈값(Mode) 계산
print("\n최빈값(Mode):")
print(df.mode())  # 각 열의 최빈값 계산

# 최대값(Max)과 최소값(Min) 계산
print("\n최댓값(Max)과 최솟값(Min):")
print("최댓값(Max):\n", df.max())  # 각 열의 최대값 계산
print("최솟값(Min):\n", df.min())  # 각 열의 최소값 계산
```

```python
# 표준편차(Standard Deviation) 계산
print("\n표준편차(Standard Deviation):")
print(df.std())  # 각 열의 표준편차 계산

# 분산(Variance) 계산
print("\n분산(Variance):")
print(df.var())  # 각 열의 분산 계산
```
